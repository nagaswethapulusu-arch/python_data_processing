# Python Data Processing
This notebook processes employee records and generates summary statistics.

```python
employees = [
    {"id": 1, "name": "Asha", "department": "Engineering", "salary": 65000, "experience": 4},
    {"id": 2, "name": "Ravi", "department": "Marketing", "salary": 48000, "experience": 2},
    {"id": 3, "name": "Meena", "department": "HR", "salary": 42000, "experience": 3},
    {"id": 4, "name": "Kiran", "department": "Engineering", "salary": 95000, "experience": 9},
    {"id": 5, "name": "Swetha", "department": "Finance", "salary": 70000, "experience": 6},
    {"id": 6, "name": "Harsh", "department": "Engineering", "salary": 58000, "experience": 1},
    {"id": 7, "name": "Rupesh", "department": "Marketing", "salary": 55000, "experience": 5},
    {"id": 8, "name": "Akhil", "department": "Finance", "salary": 88000, "experience": 8},
    {"id": 9, "name": "Divya", "department": "HR", "salary": 46000, "experience": 2},
    {"id": 10, "name": "Teja", "department": "Marketing", "salary": 62000, "experience": 7},
]
```

```python
def get_level(years):
    if years < 3:
        return "Junior"
    elif years <= 6:
        return "Mid"
    else:
        return "Senior"

def process_record(employee):
    processed = employee.copy()
    processed["level"] = get_level(employee["experience"])
    processed["bonus"] = employee["salary"] * 0.10
    return processed

def process_all(records):
    result = []
    for employee in records:
        result.append(process_record(employee))
    return result
```

```python
def summary_statistics(records):
    salaries = [r["salary"] for r in records]
    highest = max(salaries)
    lowest = min(salaries)
    return {
        "count": len(records),
        "total": sum(salaries),
        "average": sum(salaries) / len(salaries),
        "highest": highest,
        "lowest": lowest,
        "top_earner": records[salaries.index(highest)]["name"],
        "lowest_earner": records[salaries.index(lowest)]["name"],
    }

def group_by_department(records):
    groups = {}
    for r in records:
        groups.setdefault(r["department"], []).append(r["salary"])
    summary = {}
    for department, salaries in groups.items():
        summary[department] = {"count": len(salaries), "average": sum(salaries) / len(salaries)}
    return summary
```

```python
processed = process_all(employees)

print("--- Processed records (first 5) ---")
print(f"{'Name':<8}{'Dept':<13}{'Salary':<8}{'Exp':<5}{'Level':<8}Bonus")
for r in processed[:5]:
    print(f"{r['name']:<8}{r['department']:<13}{r['salary']:<8}{r['experience']:<5}{r['level']:<8}{r['bonus']:,.0f}")

stats = summary_statistics(processed)
print()
print("--- Summary statistics ---")
print(f"Total records: {stats['count']}")
print(f"Total salary: Rs.{stats['total']:,}")
print(f"Average salary: Rs.{stats['average']:,.2f}")
print(f"Highest salary: Rs.{stats['highest']:,} ({stats['top_earner']})")
print(f"Lowest salary: Rs.{stats['lowest']:,} ({stats['lowest_earner']})")

print()
print("--- Department summary ---")
for department, info in group_by_department(processed).items():
    print(f"{department:<13}{info['count']} employees, average Rs.{info['average']:,.2f}")

print()
print("--- Level count ---")
levels = {}
for r in processed:
    levels[r["level"]] = levels.get(r["level"], 0) + 1
for level, count in levels.items():
    print(f"{level}: {count}")
```
